# Show DataSet Requirement Check

```python
import pandas as pd

#  LOAD DATASET

nepal_file = "nepal.csv"

df = pd.read_csv(nepal_file)

#BASIC DATASET INFORMATION

rows = df.shape[0]
columns = df.shape[1]

# IDENTIFY NUMERICAL & CATEGORICAL COLUMNS

numerical_columns = df.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# 4. ASSIGNMENT REQUIREMENT CHECK

has_10000_records = rows > 10000
has_numerical = len(numerical_columns) > 0
has_categorical = len(categorical_columns) > 0

assignment_pass = (
    has_10000_records
    and has_numerical
    and has_categorical
)

# 5. DISPLAY RESULTS


print("        DATASET REQUIREMENT CHECK")


print(f"\nDataset File       : {nepal_file}")
print(f"Total Records     : {rows:,}")
print(f"Total Columns     : {columns}")

# NUMERICAL ATTRIBUTES

print("\n")
print("  NUMERICAL ATTRIBUTES")


print(f"Count: {len(numerical_columns)}")


# CATEGORICAL ATTRIBUTES


print("  CATEGORICAL ATTRIBUTES")


print(f"Count: {len(categorical_columns)}")


# ASSIGNMENT REQUIREMENTS

print("\n")
print("        ASSIGNMENT REQUIREMENTS")


print(
    f"\nRecords > 10,000       : "
    f"{'PASS' if has_10000_records else 'FAIL'}"
)

print(
    f"Numerical Attributes   : "
    f"{'PASS' if has_numerical else 'FAIL'}"
)

print(
    f"Categorical Attributes : "
    f"{'PASS' if has_categorical else 'FAIL'}"
)


```

# **Import Libraries + Load Dataset**

```python
import pandas as pd
import numpy as np
import plotly.express as px


df = pd.read_csv("nepal.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
import re
import os

def clean_column_name(col):
    col = str(col).strip().lower()
    col = re.sub(r'[^a-z0-9]+', '_', col)
    col = re.sub(r'_+', '_', col)
    col = col.strip('_')
    return col

df.columns = [clean_column_name(col) for col in df.columns]


new_columns = []
seen = {}

for col in df.columns:
    if col not in seen:
        seen[col] = 0
        new_columns.append(col)
    else:
        seen[col] += 1
        new_columns.append(f"{col}_{seen[col]}")

df.columns = new_columns

print("\n Cleaned columns:")
print(df.columns.tolist())

print("\nDataset shape:")
print(df.shape)

print("\nDuplicate column names:")
print(df.columns.duplicated().sum())


df.head()
```

# AutoViz

```python


df = pd.read_csv("nepal.csv")

print("Original dataset:", df.shape)

autoviz_data = df.copy()

autoviz_data.columns = [
    f"feature_{i+1:02d}"
    for i in range(len(autoviz_data.columns))
]


from autoviz.AutoViz_Class import AutoViz_Class

AV = AutoViz_Class()

print("Dataset:", autoviz_data.shape)

print()

autoviz_result = AV.AutoViz(
    filename="",
    sep=",",
    depVar="",
    dfte=autoviz_data,
    header=0,
    verbose=1,
    lowess=False,
    chart_format="html",
    max_rows_analyzed=150000,
    max_cols_analyzed=30
)

print()

print(" AUTOVIZ COMPLETED!")
```

# Create DATAPREP

```python
# CREATE DATAPREP REPORT

import subprocess

dataprep_code = r'''
import pandas as pd
from dataprep.eda import create_report

df = pd.read_csv("/content/nepal.csv")

print("Dataset shape:", df.shape)
print("Creating DataPrep report...")

report = create_report(df)

report.save("/content/dataprep_report.html")

print("DATAPREP_SUCCESS")
'''

with open("/content/run_dataprep.py", "w") as f:
    f.write(dataprep_code)

result = subprocess.run(
    [
        "/content/dataprep_env/bin/python",
        "/content/run_dataprep.py"
    ],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print("\n--- ERROR/INFO ---")
    print(result.stderr)
```
